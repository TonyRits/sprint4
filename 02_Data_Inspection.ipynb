## Setup — Load the dataset

### Concept
Before inspecting data, load the intended file and confirm that it loaded successfully. Pandas represents tabular data using a `DataFrame`.

### Example
A telecom analyst loads customer records before checking customer tenure, contract types, charges, and churn labels.

### Code

```python
import pandas as pd
import numpy as np
from pathlib import Path

DATA_PATH = Path("../dataset/WA_Fn-UseC_-Telco-Customer-Churn.csv")

# Also support running this notebook from the project root.
if not DATA_PATH.exists():
    alternative = Path("dataset/WA_Fn-UseC_-Telco-Customer-Churn.csv")
    if alternative.exists():
        DATA_PATH = alternative

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"CSV not found: {DATA_PATH.resolve()}. "
        "Download the dataset and update DATA_PATH."
    )

df = pd.read_csv(DATA_PATH)
print("Dataset loaded successfully.")
display(df.head())
```

### Output explanation
You should see the first few customer records. If the file is not found, check the filename and folder location.
Confirming the correct file was loaded prevents the rest of the analysis from being performed on an unintended dataset.

## 1. Number of Rows

### Concept
A row usually represents one observation. In this dataset, one row is intended to represent one customer record.

### Example
If there are 7,043 rows, the file contains 7,043 customer records; this does not automatically guarantee 7,043 unique customers.

### Code

```python
print("Number of rows:", len(df))
print("Number of rows using shape:", df.shape[0])
```

### Output explanation
Both expressions should return the same number because `len(df)` and `df.shape[0]` count DataFrame rows.
The row count is a baseline to compare against later checks for duplicates, missing target values, and any records that may require special attention.

## 2. Number of Columns

### Concept
Columns represent attributes or fields recorded for each observation.

### Example
A customer dataset may include tenure, contract type, monthly charges, and churn status.

### Code

```python
print("Number of columns:", len(df.columns))
print("Column names:")
print(df.columns.to_list())
```

### Output explanation
The first line shows the number of columns; the list shows their exact names.
Column names help identify fields that may represent features, identifiers, or the target. Exact spelling matters when selecting columns in code.

## 3. Column Names

### Concept
Column names are labels used to reference fields. Clear names make analysis easier, but names alone do not fully explain a field's meaning.

### Example
`MonthlyCharges` suggests a recurring monthly charge, while `Churn` suggests whether the customer left.

### Code

```python
for position, column in enumerate(df.columns, start=1):
    print(f"{position:02d}. {column}")
```

### Output explanation
Each column is displayed with its position in the dataset.
Use the source documentation to confirm definitions and units, especially for fields whose meaning is not obvious from the name.

## 4. Dataset Index

### Concept
A pandas index labels rows in a DataFrame. When reading a CSV without specifying an index column, pandas usually creates a default integer index starting at zero.

### Example
Index label `0` refers to the first row currently in the DataFrame. It is not necessarily a business identifier such as `customerID`.

### Code

```python
print("Index object:", df.index)
print("First five index labels:", df.index[:5].to_list())
print("Is index unique?", df.index.is_unique)
```

### Output explanation
The output describes the DataFrame's row index, shows its first labels, and checks whether the index labels are unique.

### Insight
A unique default index does not prove that customer records are unique. Check business identifiers and duplicated rows separately.

### Example
An analyst checks the first five records after importing a CSV.

```python
df.head()
```

## 6. `tail()`

### Concept
`tail(n)` returns the last `n` rows. It can help reveal issues near the end of a file that a first-row preview might miss.

### Example
Check the last five records to see whether the file ends with valid data rows.

Compare the beginning and end of the file for unusual formatting or unexpected records. This is only a quick visual check, not a complete validation.

## 7. `sample()`

### Concept
`sample(n)` selects a random sample of rows. It can give a quick view of records from different parts of the dataset.

### Example
A reviewer inspects five randomly selected customers rather than only the first or last records.

### Code

```python
df.sample(n=min(5, len(df)), random_state=42)
```

### Output explanation
Up to five rows are displayed. `random_state=42` makes the selected sample reproducible when the data is unchanged.
Random sampling can reveal values not seen in `head()` or `tail()`, but a small sample cannot establish that the entire dataset is clean.

## 9. `info()`

### Concept
`info()` prints a structural summary, including column names, non-null counts, data types, and memory usage.

### Example
A charge field may appear as `object` instead of a numeric type if it contains blank strings or other text.

### Code

```python
df.info()
```

### Output explanation
For each column, compare its non-null count with the total number of rows. The dtype column indicates how pandas currently stores values.
A lower non-null count can indicate missing values recognized by pandas. However, empty strings and placeholders may not count as missing until investigated. A numeric-looking text column also needs attention.

## 10. `describe()`

### Concept
`describe()` summarizes data using statistics. By default, it summarizes numerical columns with count, mean, standard deviation, minimum, quartiles, and maximum.

### Example
For `MonthlyCharges`, the summary can show the typical charge, spread, and range.

### Code

```python
display(df.describe())
```

### Output explanation
- `count`: non-missing numeric values
- `mean`: arithmetic average
- `std`: standard deviation
- `min` and `max`: smallest and largest values
- `25%`, `50%`, `75%`: quartiles; the 50th percentile is the median

Unexpected minima, maxima, or large gaps between the mean and median may warrant further inspection. Summary statistics do not explain why a value is unusual.

## 11. `dtypes`

### Concept
The `dtypes` attribute shows the pandas data type assigned to each column.

### Example
A column stored as `object` might contain category labels, identifiers, or numbers stored as text.

### Code

```python
display(df.dtypes.to_frame(name="data_type"))
```

### Output explanation
Each column is paired with its current pandas dtype, such as `int64`, `float64`, or `object`.
Dtypes describe storage, not necessarily business meaning. For example, `SeniorCitizen` may be stored as an integer but represent a binary category.

## 12. `nunique()`

### Concept
`nunique()` counts distinct values in each column. By default, missing values are excluded.

### Example
A column with two unique non-missing values may be binary, while a customer ID may have nearly as many unique values as rows.

### Code

```python
unique_counts = df.nunique()
display(unique_counts.to_frame(name="unique_non_missing_values").sort_values(
    "unique_non_missing_values", ascending=False
))
```

### Output explanation
The table ranks columns by the number of distinct non-missing values.
High cardinality may indicate an identifier or a field with many distinct values. Low cardinality may indicate a binary or categorical feature. Always confirm using column meaning.

## 13. `value_counts()`

### Concept
`value_counts()` counts how often each value appears in a Series. It is especially useful for categorical columns and target labels.

### Example
Counting `Contract` values can show whether month-to-month contracts are more common than one-year or two-year contracts.

### Code

```python
for column in ["Contract", "InternetService", "PaymentMethod", "Churn"]:
    if column in df.columns:
        print(f"\nValue counts for {column}:")
        display(df[column].value_counts(dropna=False).to_frame("count"))
```

### Output explanation
Each table lists observed values and their counts. `dropna=False` includes missing values if pandas recognizes any.
Counts can reveal dominant categories, rare categories, unexpected labels, and possible class imbalance. Later notebooks will explore these patterns more deeply.

## 14. Identify Column Roles

### Concept
Before deeper EDA, classify columns by their likely role: numerical, categorical/text, date/time, identifier, or potential target. This classification is a starting point and should be verified against business definitions.

### Example
`customerID` is a likely identifier; `Churn` is the target; `MonthlyCharges` is a numerical measurement; `Contract` is categorical.

### Code

```python

numeric_columns = df.select_dtypes(include=np.number).columns.to_list()

categorical_columns = df.select_dtypes(
    include=["object", "str", "category"]
).columns.to_list()

datetime_columns = df.select_dtypes(
    include=["datetime", "datetimetz"]
).columns.to_list()

identifier_columns = [
    col for col in df.columns
    if col.lower() in {"customerid", "customer_id", "id", "record_id"}
    or col.lower().endswith("_id")
]

potential_target_columns = [
    col for col in [
        "Churn", "Exited", "Attrition",
        "target", "Target", "label"
    ]
    if col in df.columns
]

print("Numerical dtype columns:", numeric_columns)
print("\nCategorical/text dtype columns:", categorical_columns)
print("\nDate/time dtype columns:", datetime_columns)
print("\nPotential identifier columns:", identifier_columns)
print("\nPotential target columns:", potential_target_columns)
```

### Output explanation
The lists group columns using their current pandas dtypes and a few name-based candidates. Name-based detection is only a heuristic.
Some fields need manual review. Numeric codes may represent categories, dates may be stored as text, and an identifier may not follow a standard naming convention.

### AI/ML relevance
Correctly identifying column roles helps determine what to inspect next and prevents unsuitable columns from being treated as ordinary numeric features.
